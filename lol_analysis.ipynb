## API Functions

The following function allows you to fetch summoner data and other information from the Riot Games API.

## Get Summoner Data

```python
# Function to get summoner data by summoner name
def get_summoner_data(summoner_name):
    url = f'{BASE_URL}/summoner/v4/summoners/by-name/{summoner_name}'
    headers = {'X-Riot-Token': API_KEY}
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code} - {response.text}")
        return None
```
